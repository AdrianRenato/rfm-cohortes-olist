```python
import pandas as pd
import zipfile
import os
```

```python
orders = pd.read_csv('/content/olist_orders_dataset.csv')
order_items = pd.read_csv('/content/olist_order_items_dataset.csv')
customers = pd.read_csv('/content/olist_customers_dataset.csv')
```

```python
orders.head()
```

```python
orders.info()
```

```python
orders.describe(include='all')
```

```python
orders_clean = orders[orders['order_status'] == 'delivered'].copy()
```

```python
orders_clean['order_purchase_timestamp'] = pd.to_datetime(orders_clean['order_purchase_timestamp'])
```

```python
orders_clean = orders_clean[['order_id', 'customer_id', 'order_purchase_timestamp']]
```

```python
order_items_clean = order_items[['order_id', 'price']].copy()
```

```python
order_items_clean = order_items_clean.groupby('order_id', as_index=False)['price'].sum()
```

```python
order_items.head()
```

```python
customers_clean = customers[['customer_id', 'customer_unique_id']].copy()
```

```python
print(orders_clean.shape)
orders_clean.head()
```

```python
print(order_items_clean.shape)
order_items_clean.head()
```

```python
print(customers_clean.shape)
customers_clean.head()
```

```python
orders_customers = orders_clean.merge(customers_clean, on='customer_id', how='inner')
```

```python
tabla_maestra = orders_customers.merge(order_items_clean, on='order_id', how='inner')
```

```python
tabla_maestra = tabla_maestra[['customer_unique_id', 'order_purchase_timestamp', 'price']]
```

```python
print(tabla_maestra.shape)
tabla_maestra.head()
```

```python
fecha_referencia = tabla_maestra['order_purchase_timestamp'].max() + pd.Timedelta(days=1)
print(fecha_referencia)
```

```python
rfm = tabla_maestra.groupby('customer_unique_id').agg(
    recencia=('order_purchase_timestamp', lambda x: (fecha_referencia - x.max()).days),
    frecuencia=('order_purchase_timestamp', 'count'),
    monetario=('price', 'sum')
).reset_index()
```

```python
print(rfm.shape)
rfm.head(10)
```

```python
rfm['r_score'] = pd.qcut(rfm['recencia'], 5, labels=[5, 4, 3, 2, 1])
```

```python
rfm['f_score'] = pd.qcut(rfm['frecuencia'].rank(method='first'), 5, labels=[1, 2, 3, 4, 5])
```

```python
rfm['m_score'] = pd.qcut(rfm['monetario'], 5, labels=[1, 2, 3, 4, 5])
```

```python
print(rfm.head(10))
```

```python
rfm['rfm_score'] = rfm['r_score'].astype(str) + rfm['f_score'].astype(str) + rfm['m_score'].astype(str)
```

```python
rfm['rfm_promedio'] = rfm[['r_score', 'f_score', 'm_score']].astype(int).mean(axis=1)
```

```python
rfm.head(10)
```

```python
def asignar_segmento(row):
    r = int(row['r_score'])
    f = int(row['f_score'])

    if r >= 4 and f >= 4:
        return 'Champions'
    elif r >= 4 and f < 4:
        return 'Nuevos / Prometedores'
    elif r < 4 and f >= 4:
        return 'Leales en riesgo'
    elif r <= 2 and f <= 2:
        return 'Perdidos'
    else:
        return 'En riesgo'

rfm['segmento'] = rfm.apply(asignar_segmento, axis=1)
```

```python
rfm.head(10)
```

```python
rfm['segmento'].value_counts()
```

```python
def score_frecuencia(f):
    if f == 1:
        return 1
    elif f == 2:
        return 2
    elif f == 3:
        return 3
    elif f in [4, 5]:
        return 4
    else:
        return 5

rfm['f_score'] = rfm['frecuencia'].apply(score_frecuencia)
```

```python
rfm['rfm_score'] = rfm['r_score'].astype(str) + rfm['f_score'].astype(str) + rfm['m_score'].astype(str)
rfm['rfm_promedio'] = rfm[['r_score', 'f_score', 'm_score']].astype(int).mean(axis=1)
rfm['segmento'] = rfm.apply(asignar_segmento, axis=1)

rfm['segmento'].value_counts()
```

```python
def asignar_segmento(row):
    r = int(row['r_score'])
    f = int(row['f_score'])
    m = int(row['m_score'])

    if f >= 3:
        return 'Recurrentes / Leales'
    elif r >= 4 and m >= 4:
        return 'Nuevos de alto valor'
    elif r >= 4:
        return 'Nuevos / Recientes'
    elif r <= 2 and m <= 2:
        return 'Perdidos'
    else:
        return 'En riesgo'

rfm['segmento'] = rfm.apply(asignar_segmento, axis=1)
```

```python
rfm['rfm_score'] = rfm['r_score'].astype(str) + rfm['f_score'].astype(str) + rfm['m_score'].astype(str)
rfm['rfm_promedio'] = rfm[['r_score', 'f_score', 'm_score']].astype(int).mean(axis=1)
rfm['segmento'] = rfm.apply(asignar_segmento, axis=1)

rfm['segmento'].value_counts()
```

```python
tabla_maestra['mes_compra'] = tabla_maestra['order_purchase_timestamp'].dt.to_period('M')
```

```python
tabla_maestra['cohorte'] = tabla_maestra.groupby('customer_unique_id')['mes_compra'].transform('min')
```

```python
tabla_maestra['periodo_cohorte'] = (tabla_maestra['mes_compra'] - tabla_maestra['cohorte']).apply(lambda x: x.n)
```

```python
tabla_maestra[['customer_unique_id', 'mes_compra', 'cohorte', 'periodo_cohorte']].head(10)
```

```python
cohorte_data = tabla_maestra.groupby(['cohorte', 'periodo_cohorte'])['customer_unique_id'].nunique().reset_index()
```

```python
cohorte_pivot = cohorte_data.pivot(index='cohorte', columns='periodo_cohorte', values='customer_unique_id')
```

```python
cohorte_pivot
```

```python
tamano_cohorte = cohorte_pivot[0]
retencion = cohorte_pivot.divide(tamano_cohorte, axis=0) * 100
```

```python
retencion.round(1)
```
